```python
job_description = """
We are looking for a Python Developer.

Required Skills:

C++
Python
MongoDB
Docker
NLP
Git
JavaScript
Deep Learning (DL)
Node.js
"""
```

```python
jd_skills = skill_chain.invoke(
    {
        "text": job_description
    }
)

print(jd_skills)
```

```python
resume_set = set(
    map(
        str.strip,
        resume_skills.split(",")
    )
)

jd_set = set(
    map(
        str.strip,
        jd_skills.split(",")
    )
)
```

```python
matched_skills = resume_set.intersection(
    jd_set
)

missing_skills = jd_set - resume_set
```

```python
ats_score = (
    len(matched_skills)
    /
    len(jd_set)
) * 100
```

```python
print(
    round(ats_score,2)
)
```

```python
print(
    "\nATS SCORE:",
    round(ats_score,2),
    "%"
)

print("\nMATCHED SKILLS")

for skill in matched_skills:
    print("✓", skill)

print("\nMISSING SKILLS")

for skill in missing_skills:
    print("✗", skill)
```
